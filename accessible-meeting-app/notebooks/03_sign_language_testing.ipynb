# Notebook 3: Sign Language Recognition Testing
Tests the vision model for sign recognition (default `dima806/asl_alphabet_image_detection`, ASL **alphabet** letters from a still image). Without an API key (or a local model) it returns an honest error - it never invents a sign.

```python
import os, sys
from dotenv import load_dotenv

# Run from the notebooks/ folder: project root is one level up.
sys.path.append('..')
load_dotenv('../.env')

from PIL import Image
import numpy as np
from services.huggingface_service import huggingface_service
from utils.config import Config

print(f"Configured Sign Model: {Config.HUGGINGFACE_SIGN_MODEL}")
```

```python
# Random noise image just exercises the pipeline. For a real test use: Image.open('hand_sign_A.jpg')
test_pil = Image.fromarray(np.random.randint(0, 255, (224, 224, 3), dtype=np.uint8))

result = huggingface_service.detect_sign_language(test_pil)
print("--- SIGN DETECTION TEST RESULT ---")
for k, v in result.items():
    print(f"{k}: {v}")
```
