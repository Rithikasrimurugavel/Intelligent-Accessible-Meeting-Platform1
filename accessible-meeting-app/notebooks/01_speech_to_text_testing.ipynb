# Notebook 1: Speech-to-Text (STT) Testing
Tests the Hugging Face STT model (default `openai/whisper-large-v3-turbo`). Without an API key the free Google Web Speech fallback is used (needs internet). Silent recordings are rejected.

```python
import os, sys
from dotenv import load_dotenv

# Run from the notebooks/ folder: project root is one level up.
sys.path.append('..')
load_dotenv('../.env')

from services.huggingface_service import huggingface_service
from utils.config import Config

print(f"Configured STT Model: {Config.HUGGINGFACE_STT_MODEL}")
print(f"HF key configured / live mode: {huggingface_service.is_live}")
```

```python
# Synthetic 1-second WAV (a tone, not speech - a live model will likely return 'no speech').
# For a real test, replace with: test_audio_bytes = open('my_recording.wav', 'rb').read()
import io, wave, struct, math

def create_dummy_wav():
    buf = io.BytesIO()
    with wave.open(buf, 'wb') as wf:
        wf.setnchannels(1); wf.setsampwidth(2); wf.setframerate(16000)
        wf.writeframes(b''.join(struct.pack('<h', int(8000 * math.sin(2 * math.pi * 440 * i / 16000))) for i in range(16000)))
    return buf.getvalue()

test_audio_bytes = create_dummy_wav()
print(f"Generated test WAV: {len(test_audio_bytes)} bytes")
```

```python
result = huggingface_service.speech_to_text(test_audio_bytes)
print("--- STT TEST RESULT ---")
for k, v in result.items():
    print(f"{k}: {v}")
```
