# Notebook 2: Text-to-Speech (TTS) Testing
Tests Hugging Face TTS (default `facebook/mms-tts-eng`) with a gTTS fallback.

```python
import os, sys
from dotenv import load_dotenv

# Run from the notebooks/ folder: project root is one level up.
sys.path.append('..')
load_dotenv('../.env')

from services.huggingface_service import huggingface_service
from utils.config import Config

print(f"Configured TTS Model: {Config.HUGGINGFACE_TTS_MODEL}")
```

```python
from IPython.display import Audio, display
from utils.helpers import detect_audio_mime

success, audio_bytes, note = huggingface_service.text_to_speech("Welcome to the Accessible Meeting App. Voice Guidance is active.")
print("--- TTS TEST RESULT ---")
print(f"Success: {success}")
print(f"Audio bytes: {len(audio_bytes) if audio_bytes else 0}  ({detect_audio_mime(audio_bytes) if audio_bytes else '-'})")
print(f"Note: {note}")
if success:
    display(Audio(audio_bytes))
```
